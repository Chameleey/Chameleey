```python
import pandas as pd
from datasets import Dataset

# Keep original dataset untouched
work_df = df.copy()

# Replace NaN values used in prompts
for col in [
    "intent", "target_type", "target_value",
    "ports", "modifiers", "tool_hint",
    "retrieved_tool", "retrieved_chunk",
    "expected_tool", "expected_command"
]:
    work_df[col] = work_df[col].fillna("NONE").astype(str)


def build_text(row):

    user_content = f"""Generate exactly one penetration-testing command.

Intent: {row['intent']}
Target type: {row['target_type']}
Target: {row['target_value']}
Ports: {row['ports']}
Modifiers: {row['modifiers']}
Tool hint: {row['tool_hint']}
Retrieved tool: {row['retrieved_tool']}
Retrieved context: {row['retrieved_chunk']}

Return only the command."""

    messages = [
        {
            "role": "user",
            "content": user_content
        },
        {
            "role": "assistant",
            "content": row["expected_command"]
        }
    ]

    return tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=False
    )


work_df["text"] = work_df.apply(build_text, axis=1)

train_df = work_df[work_df["split"] == "train"].copy()
val_df   = work_df[work_df["split"] == "validation"].copy()
test_df  = work_df[work_df["split"] == "test"].copy()

train_dataset = Dataset.from_pandas(
    train_df[["text"]],
    preserve_index=False
)

val_dataset = Dataset.from_pandas(
    val_df[["text"]],
    preserve_index=False
)

test_dataset = Dataset.from_pandas(
    test_df[["text"]],
    preserve_index=False
)

print("TRAIN:", len(train_dataset))
print("VALIDATION:", len(val_dataset))
print("TEST:", len(test_dataset))

print("\n===== SAMPLE GEMMA TRAINING EXAMPLE =====\n")
print(train_dataset[0]["text"])
```

```python
import pandas as pd
from datasets import Dataset

train_df = pd.read_csv("c2_sft_train.csv")
val_df   = pd.read_csv("c2_sft_validation.csv")
test_df  = pd.read_csv("c2_sft_test.csv")

train_safe = train_df[train_df["safety_passed"] == True].copy()
val_safe   = val_df[val_df["safety_passed"] == True].copy()
test_safe  = test_df[test_df["safety_passed"] == True].copy()

def build_training_text(row):
    user_prompt = f"""Generate exactly one penetration-testing command.

Intent: {row['intent']}
Target: {row['target']}
Ports: {row['ports']}
Modifiers: {row['modifiers']}
Tool: {row['tool']}

Return only the command."""

    messages = [
        {"role": "user", "content": user_prompt},
        {"role": "assistant", "content": str(row["expected_command"])}
    ]

    return tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=False
    )

train_safe["text"] = train_safe.apply(build_training_text, axis=1)
val_safe["text"]   = val_safe.apply(build_training_text, axis=1)
test_safe["text"]  = test_safe.apply(build_training_text, axis=1)

train_dataset = Dataset.from_pandas(train_safe[["text"]], preserve_index=False)
val_dataset   = Dataset.from_pandas(val_safe[["text"]], preserve_index=False)
test_dataset  = Dataset.from_pandas(test_safe[["text"]], preserve_index=False)

print("Train:", len(train_dataset))
print("Validation:", len(val_dataset))
print("Test:", len(test_dataset))
print("\nSample:\n")
print(train_dataset[0]["text"])
```

```python
import torch
from peft import LoraConfig, get_peft_model

model.gradient_checkpointing_enable()
model.config.use_cache = False

lora_config = LoraConfig(
    r=16,
    lora_alpha=32,
    target_modules=["q_proj", "v_proj"],
    lora_dropout=0.05,
    bias="none",
    task_type="CAUSAL_LM"
)

model = get_peft_model(model, lora_config)

print("✅ LoRA adapter attached")
model.print_trainable_parameters()

print(
    "GPU memory allocated:",
    round(torch.cuda.memory_allocated() / 1024**3, 2),
    "GB"
)
```

```python
import torch

sample_text = train_dataset[0]["text"]

batch = tokenizer(
    sample_text,
    return_tensors="pt",
    truncation=True,
    max_length=384
)

batch = {k: v.to(model.device) for k, v in batch.items()}
batch["labels"] = batch["input_ids"].clone()

model.train()

outputs = model(**batch)
loss = outputs.loss

print("Loss before backward:", float(loss))

loss.backward()

print("✅ 1-step forward/backward test successful")
print("GPU memory allocated:", round(torch.cuda.memory_allocated() / 1024**3, 2), "GB")
print("GPU max memory allocated:", round(torch.cuda.max_memory_allocated() / 1024**3, 2), "GB")

model.zero_grad(set_to_none=True)
```

```python
from transformers import TrainingArguments, Trainer, DataCollatorForLanguageModeling

MAX_LENGTH = 384

def tokenize_function(examples):
    return tokenizer(
        examples["text"],
        truncation=True,
        max_length=MAX_LENGTH,
        padding=False
    )

tokenized_train = train_dataset.map(
    tokenize_function,
    batched=True,
    remove_columns=["text"]
)

tokenized_val = val_dataset.map(
    tokenize_function,
    batched=True,
    remove_columns=["text"]
)

data_collator = DataCollatorForLanguageModeling(
    tokenizer=tokenizer,
    mlm=False
)

training_args = TrainingArguments(
    output_dir="/content/neuroshell-c2-gemma-lora",

    num_train_epochs=1,

    per_device_train_batch_size=1,
    per_device_eval_batch_size=1,
    gradient_accumulation_steps=4,

    learning_rate=2e-4,
    weight_decay=0.01,
    warmup_ratio=0.05,

    fp16=True,

    logging_steps=10,

    eval_strategy="steps",
    eval_steps=40,

    save_strategy="steps",
    save_steps=40,
    save_total_limit=2,

    load_best_model_at_end=True,

    report_to="none"
)

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=tokenized_train,
    eval_dataset=tokenized_val,
    data_collator=data_collator
)

print("✅ TRAINER READY")
print("Training examples:", len(tokenized_train))
print("Validation examples:", len(tokenized_val))
print("Epochs:", training_args.num_train_epochs)
print("Effective batch size:",
      training_args.per_device_train_batch_size *
      training_args.gradient_accumulation_steps)
```

```python
print("pad_token:", tokenizer.pad_token)
print("pad_token_id:", tokenizer.pad_token_id)
print("eos_token:", tokenizer.eos_token)
print("eos_token_id:", tokenizer.eos_token_id)
```

```python
tokenizer.pad_token = tokenizer.eos_token
tokenizer.pad_token_id = tokenizer.eos_token_id

print("✅ pad token set to EOS")
```

```python
train_result = trainer.train()
```

```python
save_path = "/content/neuroshell-c2-gemma-lora"

trainer.save_model(save_path)
tokenizer.save_pretrained(save_path)

print("✅ NeuroShell C2 LoRA model saved!")
print("Saved to:", save_path)
```
